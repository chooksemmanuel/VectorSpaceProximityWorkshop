<a id="tokenizer"></a><a id="normalization"></a>

### Tokenizer & Normalization

Convert text into a list of words (tokens)

```python
import re

def my_tokenizer(text):
    
    # Lowercase: Change "The" to "the"
    lowered_text = str(text).lower()
    
    # Remove punctuation: Remove characters like ! . , ?
    # [^a-z0-9\s] means "everything that is NOT a letter, number, or space"
    clean_text = re.sub(r'[^a-z0-9\s]', '', lowered_text)
    
    # Split: cut the sentence into single words.
    tokens = clean_text.split()
    
    return tokens

# --- Test---
example_sentence = "Warning: CVE-2024! Ransomware is attacking Windows."
print("Original :", example_sentence)
print("Tokens :", my_tokenizer(example_sentence))
```

<a id="stopwordremoval"></a>

### Stop-word removal

Remove words that have no special meaning.

```python
import nltk
from nltk.corpus import stopwords

# Download the list nltk
nltk.download('stopwords')

# 2. Get the English stop words list (set)
stop_words = set(stopwords.words('english'))

def remove_stopwords(tokens):
    
    # Keep the word ONLY IF it is not in the stop_words list
    filtered_tokens = [word for word in tokens if word not in stop_words]
    
    return filtered_tokens

# --- Test ---
# We use the tokens from the previous step
sample_tokens = ["the", "system", "is", "under", "a", "ransomware", "attack"]
print("Before :", sample_tokens)
print("After :", remove_stopwords(sample_tokens))
```

<a id="stemmingorlemmatization"></a>

### Stemming or lemmatization

| Feature| Stemming | Lemmatization  |
| :--- | :--- | :--- |
Method |Heuristic rules |Dictionary lookup|
Speed |Very Fast |Slower |
Accuracy |Lower: May create non-real words (e.g., "vulner"). |Higher: Always produces real dictionary words. |
Example |"studies" $\rightarrow$ "studi"|"studies" $\rightarrow$ "study"|

```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download necessary data for Lemmatization
nltk.download('wordnet')

# Initialize tools
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# --- Function for Stemming ---
def apply_stemming(tokens):
    # Cut words
    return [stemmer.stem(word) for word in tokens]

# --- Function for Lemmatization ---
def apply_lemmatization(tokens):
    # Dictionary lookup
    return [lemmatizer.lemmatize(word) for word in tokens]

# --- Test and Compare ---
sample_tokens = ["exploited", "exploits", "exploiting", "vulnerabilities", "feet", "was"]

print(f"{'Original':<15} | {'Stemming':<15} | {'Lemmatization':<15}")
print("-" * 50)

for word in sample_tokens:
    s = stemmer.stem(word)
    l = lemmatizer.lemmatize(word)
    print(f"{word:<15} | {s:<15} | {l:<15}")
```

```python
# Create a final function for preprocessing
def final_preprocess(text):
    # Tokenize & Lowercase 
    tokens = my_tokenizer(text) 
    
    # Remove Stopwords 
    tokens = remove_stopwords(tokens)
    
    # Lemmatization
    # Using Lemmatizer to get the real root words
    clean_tokens = [lemmatizer.lemmatize(word) for word in tokens]
    
    return " ".join(clean_tokens)

# Apply to the 'shortDescription' column
df['shortDescription_new'] = df['shortDescription'].apply(final_preprocess)

# Show the result
df[['shortDescription', 'shortDescription_new']].head()
```

### Term frequency (TF)

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

# Use CountVectorizer (Default binary=False)
tf_vectorizer = CountVectorizer()

# Create the TF matrix
tf_matrix = tf_vectorizer.fit_transform(df['shortDescription_new'])

# Create a DataFrame to show the counts 1, 2, 3...
df_tf = pd.DataFrame(
    tf_matrix.toarray(), 
    columns=tf_vectorizer.get_feature_names_out(),
    index=df['cveID']
)

# Show the first 5 rows and 10 words
print("Term Frequency (TF) Matrix - Word Counts:")
display(df_tf.iloc[:5, :10])
```

<a id="logfrequencyweighting"></a>

### Log-frequency weighting (Log TF)

Raw TF is not ideal — a term appearing 10 times is not necessarily 10× more important than one appearing once.

**Log TF** dampens the effect of high-frequency terms:

$$w(t, d) = \begin{cases} 1 + \log_{10}(tf(t,d)) & \text{if } tf(t,d) > 0 \\ 0 & \text{otherwise} \end{cases}$$

```python
import numpy as np

# Apply the formula: 1 + log10(tf) if tf > 0, else 0
def compute_log_tf(tf):
    if tf > 0:
        # We use np.log10 for the base 10 log
        return 1 + np.log10(tf)
    else:
        return 0

# Apply the function to our TF Matrix
# .applymap() moves through every single cell in the table

#df_log_tf = df_tf.applymap(compute_log_tf) # old Pandas ver. <2.1.0
df_log_tf = df_tf.map(compute_log_tf)

# Show the difference
print("Log-frequency Weighting Matrix (Log TF):")
df_log_tf.iloc[:5, :10]
```

<a id="documentfrequency"></a>

### Document frequency (DF)

Definition:How many documents (CVE records) have this word?

```python
# Check if the word exists (1) or not (0)
df_exists = (df_tf > 0)

# Sum up the rows for each column (each word)
# axis=0 means "sum vertically"
document_frequency = df_exists.sum(axis=0)

# Create a DataFrame
df_df_results = pd.DataFrame(document_frequency, columns=['DF_Count'])

# Show the Top 10 words
print("Document Frequency (DF) - Top 10:")
df_df_results.sort_values(by='DF_Count', ascending=False).head(10)
```

<a id="inversedocumentfrequency"></a>

### Inverse document frequency (IDF)

**IDF** penalises terms that appear in many documents and rewards rare, informative terms:

$$idf_t = \log\left(\frac{N+1}{df_t+1}\right) + 1$$

- Common terms (high DF) → low IDF → less weight
- Rare terms (low DF) → high IDF → more weight

```python
import numpy as np

# Total number of documents (N)
N = len(df)

# Calculate IDF
# idf = log((N+1) / (df+1)) + 1
idf = np.log((N + 1) / (document_frequency + 1)) + 1

# Create a DataFrame
df_idf_results = pd.DataFrame(idf, columns=['IDF_Score'])

# Show the results: Rare words vs. Common words

print("IDF Scores - Top 10 Highest :")
display(df_idf_results.sort_values(by='IDF_Score', ascending=False).head(10))

print("\nIDF Scores - Top 10 Lowest :")
display(df_idf_results.sort_values(by='IDF_Score', ascending=True).head(10))
```

<a id="tfidfweighting"></a>

### TF-IDF weighting

**TF-IDF** combines both steps:

$$\text{TF-IDF}(t, d) = tf(t, d) \times idf_t$$

A term scores high only when it is **frequent in a specific document** AND **rare across the corpus** — exactly what makes it a good descriptor of that document's topic.

```python
# Calculate the final TF-IDF scores
# Log-TF X IDF
# axis=1 代表將 IDF 向量的每個值對應到矩陣的「欄位 (Words)」進行計算
df_tfidf = df_log_tf.multiply(idf, axis=1)

# Show the result
sample_cve = df_tfidf.iloc[0]
print(f"Top keywords for CVE {df_tfidf.index[0]}:")
print(sample_cve.sort_values(ascending=False).head(10))

# 3. Show a small part of the final matrix
print("\nFinal TF-IDF Matrix (Sample):")
display(df_tfidf.iloc[:5, 0:10])
```

<a id="cosinesimilarityretrieval"></a>

### Cosine similarity retrieval

**Cosine similarity** measures the angle between two vectors in the TF-IDF space:

$$\cos(\vec{q}, \vec{d}) = \frac{\vec{q} \cdot \vec{d}}{\|\vec{q}\| \cdot \|\vec{d}\|}$$

- Score of **1.0** = identical direction (very similar)
- Score of **0.0** = orthogonal (no shared vocabulary)

```python
from sklearn.metrics.pairwise import cosine_similarity

# query: user input
# tfidf_matrix: df_tfidf
# vectorizer:  CountVectorizer
# idf_vector:  idf Score
# top_n : show top n data

def search_vulnerability(query, tfidf_matrix, vectorizer, idf_vector, top_n=5):

    # Preprocess the query
    processed_query = final_preprocess(query)
    
    # Transform query to TF vector
    query_tf = vectorizer.transform([processed_query]).toarray()
    
    # Apply Log-TF and Multiply by IDF to get Query TF-IDF
    query_log_tf = np.where(query_tf > 0, 1 + np.log10(query_tf, where=query_tf>0), 0)
    # Log-TF X IDF 
    query_tfidf = query_log_tf * idf_vector.values
    
    # Calculate Cosine Similarity between query and all documents
    # cosine_similarity feeback is  array of arrays，so use [0]
    similarities = cosine_similarity(query_tfidf, tfidf_matrix)[0]
    
    # Get the Top N results
    results_idx = similarities.argsort()[-top_n:][::-1]
    
    # creat DataFrame
    results = []
    for idx in results_idx:
        results.append({
            "CVE_ID": tfidf_matrix.index[idx],
            "Score": similarities[idx],
            "Description": df.iloc[idx]['shortDescription']
        })
    
    return pd.DataFrame(results)

# --- Test ---
user_query_1 = "Apple iOS"
search_results = search_vulnerability(user_query_1, df_tfidf, tf_vectorizer, idf)

print(f"Search Results for: '{user_query_1}'")
display(search_results)
```

```python
user_query_2 = "Banana Android"
search_results = search_vulnerability(user_query_2, df_tfidf, tf_vectorizer, idf)

print(f"Search Results for: '{user_query_2}'")
display(search_results)
```

### Comparison Define Functions

```python
def comparison(query, top_n=5):
    modes = ['binary', 'raw_tf', 'tfidf']
    all_results = {}
    
    # Preprocess Query
    processed_query = final_preprocess(query)
    query_tf = tf_vectorizer.transform([processed_query]).toarray()
    
    for mode in modes:
        if mode == 'binary':
            # Binary: 0 / 1
            q_vec = np.where(query_tf > 0, 1, 0)
            matrix = df_incidence.values
        elif mode == 'raw_tf':
            # Raw TF: Use times
            q_vec = query_tf
            matrix = df_tf.values
        else:
            # TF-IDF: Log-TF * IDF
            query_log_tf = np.where(query_tf > 0, 1 + np.log10(query_tf, where=query_tf>0), 0)
            q_vec = query_log_tf * idf.values
            matrix = df_tfidf.values

        # Cosine Similarity
        sims = cosine_similarity(q_vec, matrix)[0]
        idx = sims.argsort()[-top_n:][::-1]
        
        res = []
        for i in idx:
            res.append({
                "Rank": len(res) + 1,
                "CVE_ID": df.iloc[i]['cveID'],
                "Score": round(sims[i], 4),
                "Description": df.iloc[i]['shortDescription']
            })
        all_results[mode] = pd.DataFrame(res)
        
    return all_results
```

## Part D — Evaluation

1. **Relevance Judgment**:
    A human looks at the search results and decides if a document is Relevant (1) or Irrelevant (0) based on the search goal.
2. **Confusion Matrix**:A table used to describe the performance of the system. It shows:
    * True Positive (TP): System found it, and it is correct.
    * False Positive (FP): System found it, but it is wrong (Spam).
    * False Negative (FN): System missed it (It was correct but not found).
    * True Negative (TN): System ignored it, and it was indeed wrong.
3. **Precision**:Out of everything the system found, how much is actually correct? (Quality)
4. **Recall**:Out of all the correct documents in the database, how many did the system find? (Quantity)
5. **F1-score**:The average  between Precision and Recall.
6. **Precision@K & @Top**:Precision calculated only for the top K results (e.g., Top 5 or Top 10).
7. **Average Precision (AP)**:This measure cares about ranking. It gives a higher score if the relevant documents are at the very top (Rank 1, 2) instead of the bottom.
8. **Mean Reciprocal Rank (MRR)**:It focuses only on the first correct answer. If the first right answer is at Rank 1, the score is $1/1 = 1$. If it is at Rank 2, the score is $1/2 = 0.5$.
9. **Cohen's Kappa Measure**:A score that shows how much two humans (User A and User B) agree on the results. If both say "this is relevant," the Kappa score is high.
10. **Smart Notation**:A 3-letter code used to describe the Term Frequency (TF) / Document Frequency (IDF) / Normalization.
    * TF: n:natural / l:logarithm: $1 + \log(tf)$
    * IDF: n:none / t:idf: $\log(N/df)$
    * Nor: n:none / c:cosine

### Query 1

```python
import numpy as np
import pandas as pd
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score, accuracy_score, cohen_kappa_score

# ==========================================================
# 1. Human Relevance Judgments (Ground Truth)
# 人工相關性判斷（地面實況）
# ==========================================================
# Based on our TF-IDF results for Query 1:
# Rank 1: CVE-2019-18426 (iPhone) -> Relevant (1)
# Rank 2: CVE-2019-7287 (Apple iOS) -> Relevant (1)
# Rank 3: CVE-2019-8526 (macOS) -> Partially Relevant (Assume 0 for strict evaluation)
# Rank 4: CVE-2024-27348 (Apache) -> Irrelevant (0)
# Rank 5: CVE-2023-42824 (Apple iOS) -> Relevant (1)

y_true = [1, 1, 0, 0, 1]  # Actual relevance (人工判斷的真實結果)
y_pred = [1, 1, 1, 1, 1]  # System prediction (系統檢索出來的都視為 Positive)

# ==========================================================
# 2. Basic Metrics Calculation (基礎指標計算)
# ==========================================================
# Precision: Quality of results | 精確率：結果的品質
precision = precision_score(y_true, y_pred)

# Recall: Completeness (Assume we found all relevant docs in top 5 for this lab)
# 召回率：完整性（假設資料庫中相關的都在這 5 筆內）
recall = recall_score(y_true, y_pred)

# F1-score: Harmonic mean of Precision and Recall | F1-分數：精確率與召回率的平衡
f1 = f1_score(y_true, y_pred)

# Accuracy: Overall correctness | 準確度：整體正確比例
accuracy = accuracy_score(y_true, y_pred)

# ==========================================================
# 3. IR Specific Metrics (資訊檢索專用指標)
# ==========================================================
# Precision@K (K=5) & @Top
p_at_k = sum(y_true) / len(y_true)

# Average Precision (AP): Considers the rank of relevant documents
# 平均精確率 (AP)：考慮相關文件出現的排名順序
def calculate_ap(labels):
    precisions = []
    relevant_found = 0
    for i, l in enumerate(labels):
        if l == 1:
            relevant_found += 1
            precisions.append(relevant_found / (i + 1))
    return np.mean(precisions) if precisions else 0

ap = calculate_ap(y_true)

# Mean Reciprocal Rank (MRR): 1 / Rank of the first relevant document
# 平均倒數排名 (MRR)：1 / 第一個正確答案的排名
first_relevant_rank = y_true.index(1) + 1
mrr = 1 / first_relevant_rank

# ==========================================================
# 4. Agreement Metric (一致性指標)
# ==========================================================
# Cohen's Kappa: Agreement between two annotators (e.g., you and a teammate)
# 柯恩卡帕係數：兩位標記人員（你與隊友）的一致性
user_a = [1, 1, 0, 0, 1] # Your labels
user_b = [1, 1, 1, 0, 1] # Teammate's labels (Simulation)
kappa = cohen_kappa_score(user_a, user_b)

# ==========================================================
# 5. Display Results (顯示結果)
# ==========================================================
print(f"--- Query 1: Evaluation Metrics ---")
print(f"Smart Notation: ltc (Log-TF, IDF, Cosine)")
print(f"-----------------------------------")
print(f"Confusion Matrix:\n{confusion_matrix(y_true, y_pred, labels=[1,0])}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-score: {f1:.4f}")
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision@5 (Top): {p_at_k:.4f}")
print(f"Average Precision (AP): {ap:.4f}")
print(f"MRR: {mrr:.4f}")
print(f"Cohen's Kappa: {kappa:.4f}")
```

### Reflection
Answer these questions:

1. Which representation worked best and why?

    TF-IDF (ltc) worked best. Unlike Binary or Raw TF, it can distinguish between "common words" and "critical keywords." In our tests, it successfully prioritized unique security terms like "Ransomware" or "iPhone" while lowering the weight of common words like "file" or "system."

2. Did TF-IDF improve over raw term counts?

    Yes, significantly. Raw TF often ranked long documents higher just because they repeated certain words many times (e.g., Query 4). TF-IDF fixed this by using Inverse Document Frequency (IDF) to penalize words that appear in too many documents, ensuring that specific vulnerabilities are ranked at the top.

3. What kinds of false positives did you observe?

    Most false positives came from "Polysemy" (words with multiple meanings). For example, in Query 3 ("Excel table"), the system retrieved a Cisco "routing table." Although the word "table" matched, the context (networking vs. spreadsheet) was completely different.

4. What kinds of relevant documents were missed?

    We missed documents that used synonyms instead of exact keywords. For instance, a query for "memory error" might miss documents that only use technical terms like "buffer overflow" or "heap corruption" without explicitly saying "memory error."

5. How did the evaluation metrics help you understand system quality?

    Metrics like MRR and AP were much more useful than simple Accuracy. They showed us that our system is excellent at ranking the best result at Rank 1, even if the overall Precision is low. This helps us understand the "User Experience" (finding the answer quickly) rather than just counting correct words.

6. How would you improve the system next?

    I would implement Stemming or Lemmatization to handle different word forms (e.g., "locking" vs "locked"). I would also add a Thesaurus/Synonym mapping to connect general terms like "memory error" to specific technical terms like "segmentation fault." Finally, using N-grams would help capture phrases like "remote code execution" as a single unit.
